```python
notas = [8.5, 7.8, 9.2, 6.9]
pesos = [0.3, 0.3, 0.2, 0.2]  # pesos das notas correspondentes
nota_ponderada = sum(nota * peso for nota, peso in zip(notas, pesos))
media_ponderada = nota_ponderada / sum(pesos)
print("A média ponderada das notas é:", media_ponderada)
```

Outra possibilidade

```python
# Notas e pesos
notas = [8.5, 7.8, 9.2, 6.9]
pesos = [3, 3, 2, 2]


print(range(len(notas)))

# Calculando o somatório do produto entre notas e pesos
somatorio_produtos = 0
for i in range(len(notas)):
    somatorio_produtos += notas[i] * pesos[i]

# Calculando o somatório dos pesos
somatorio_pesos = 0
for peso in pesos:
    somatorio_pesos += peso

# Calculando a média ponderada
media_ponderada = somatorio_produtos / somatorio_pesos

# Exibindo o resultado
print(f"A média ponderada é: {media_ponderada:.2f}")
```
