```python
df = pd.read_csv("https://raw.githubusercontent.com/INRIA/scikit-learn-mooc/main/datasets/penguins.csv")
df.head()
```

```python
df.Species.value_counts()
```

### Notas

- `.unique()` e iterações
- `df.field_A` = `df["field_A"]` = `cpo ="field_A"` + `df[cpo]`
- Solução alternativa com `groupby()`

```python
for specie in df.Species.unique():
    print(f'{specie},: {df[ df.Species == specie ]["Body Mass (g)"].mean():.2f}')
```

```python
df.groupby("Species")["Body Mass (g)"].mean()
```
