## Faixa Etária (idade) dos Colaboradores

A Faixa Etária dos colaboradores é relativamente alta com idade média dos colaboradores de 46 anos, idade mínima de 25 e máxima de 80 anos.

```python
from datetime import date
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate age from 'Data Nascimento'
df['Idade'] = df['Data Nascimento'].apply(lambda x: (date.today().year - x.year) - ((date.today().month, date.today().day) < (x.month, x.day)))

# Filter for active employees
df_ativos = df[df['Situação Colaborador'] == 'Ativo']

# Plot the distribution of ages for active employees
plt.figure(figsize=(10, 6))
sns.histplot(df_ativos['Idade'], kde=True, bins=20)
plt.title('Distribuição da Idade dos Funcionários Ativos')
plt.xlabel('Idade')
plt.ylabel('Número de Funcionários')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
