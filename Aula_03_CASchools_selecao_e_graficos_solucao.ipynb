````python
df = pd.read_csv("https://raw.githubusercontent.com/INRIA/scikit-learn-mooc/main/datasets/penguins.csv")
df['Species'] = df['Species'].str.split(' ').str[0]
df = df.dropna(subset=['Body Mass (g)'])
df.head()
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(df['Body Mass (g)'], kde=True)
plt.title('Overall Distribution of Penguin Body Mass')
plt.axvline(df['Body Mass (g)'].mean(), color='k', linestyle='dashed', linewidth=1)
plt.axvline(df['Body Mass (g)'].median(), color='r', linestyle='dashed', linewidth=1)
plt.xlabel('Body Mass (g)')
plt.ylabel('Frequency')
plt.legend(['hist', 'mean', 'median'])
plt.show()

sns.histplot(data=df, x='Body Mass (g)', hue='Species', kde=True, multiple='stack')
plt.title('Distribution of Penguin Body Mass by Species')
plt.xlabel('Body Mass (g)')
plt.ylabel('Frequency')
plt.legend(title='Species', labels=df['Species'].unique())
plt.show()
````

````python
sns.boxplot(data=df[ df.Sex != '.'], y='Body Mass (g)', hue='Sex')
plt.title('Distribution of Penguin Body Mass by Sex')
plt.xlabel('Body Mass (g)')
plt.ylabel('Frequency')
plt.show()
````

````python
sns.boxplot(data=df, y='Body Mass (g)')
plt.title('Distribution of Penguin Body Mass and Outliers')
plt.xlabel('Body Mass (g)')
plt.ylabel('Frequency')
plt.show()

sns.boxplot(data=df, y='Body Mass (g)', hue='Species')
plt.title('Distribution of Penguin Body Mass by Species')
plt.xlabel('Body Mass (g)')
plt.ylabel('Frequency')
plt.show()
````

````python
def find_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers

````

````python
find_outliers(df, 'Body Mass (g)')
````

````python
find_outliers(df[ df.Species == "Chinstrap"], 'Body Mass (g)')
````
