> **O Teorema Central do Limite (TCL)** A soma de $N$ variáveis aleatórias independentes $X$, com qualquer distribuição e variâncias semelhantes, é uma variável com distribuição que se aproxima da distribuição normal quando $N$ aumenta.

# According to the Central Limit Theorem (CLT), the distribution of sample means
# will tend towards a normal distribution, regardless of the original distribution
# of the individual random values, as the sample size (series_length in this case)
# increases. Even with a relatively small sample size of 100, the distribution
# of the means should already show a bell-shaped curve characteristic of a normal distribution.

-->


```python
num_series = 1000
series_length = 100
means = []

for _ in range(num_series):
  random_series = np.random.rand(series_length)
  means.append(np.mean(random_series))

# Plot the distribution of the means
plt.figure(figsize=(8, 6))
sns.histplot(means, bins=20, kde=True)
plt.title('Distribution of 100 Means (each from 100 random values)')
plt.xlabel('Mean Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

# PDF, CDF, PPF

```python
def plot_distributions(mu, sigma, x0):

  import numpy as np
  import matplotlib.pyplot as plt
  import seaborn as sns
  from scipy.stats import norm

  # Define os parâmetros da distribuição normal
  # mu = 1.7  # Média
  # sigma = 0.1  # Desvio padrão

  # Gera pontos para o eixo x
  x = np.linspace(mu - 4*sigma, mu + 4*sigma, 500)

  # Calcula os valores das funções RVS, PDF, CDF e PPF
  rvs_values = norm.rvs(loc=mu, scale=sigma, size=100)
  pdf_values = norm.pdf(x, loc=mu, scale=sigma)
  cdf_values = norm.cdf(x, loc=mu, scale=sigma)

  probabilities = np.linspace(0.001, 0.999, 500) # Evita 0 e 1 para ppf
  ppf_values = norm.ppf(probabilities, loc=mu, scale=sigma)

  # Cria a figura e os subplots
  fig, ax = plt.subplots(1,3,figsize=(12, 4))

  # Gráfico da PDF
  ax[0].plot(x, pdf_values, label='PDF (norm.pdf)')
  sns.kdeplot(rvs_values, label='RVS (norm.rvs)', color='red', lw=1, ax=ax[0])
  ax[0].set_title('Função Densidade de Probabilidade (PDF)')
  ax[0].set_xlabel('Valor (x)')
  ax[0].set_ylabel('Densidade de Probabilidade')
  ax[0].legend(loc='upper left')

  # Gráfico da CDF
  ax[1].plot(x, cdf_values, label='CDF (norm.cdf)', color='orange')
  ax[1].set_title('Função Distribuição Acumulada (CDF)')
  ax[1].set_xlabel('Valor (x)')
  ax[1].set_ylabel('Probabilidade Acumulada')
  ax[1].legend(loc='upper left')

  # Gráfico da PPF
  # Para a PPF, plotamos os quantis (valores x) correspondentes às probabilidades
  ax[2].plot(probabilities, ppf_values, label='PPF (norm.ppf)', color='green')
  ax[2].set_title('Função Quantil (PPF)')
  ax[2].set_xlabel('Probabilidade')
  ax[2].set_ylabel('Quantil (x)')
  ax[2].legend(loc='upper left')

  # x0 = 1.6

  pdf_x0 = norm.pdf(x0, loc=mu, scale=sigma)
  cdf_x0 = norm.cdf(x0, loc=mu, scale=sigma)

  prob_x0 = norm.cdf(x0, loc=mu, scale=sigma)
  ppf_x0 = norm.ppf(prob_x0, loc=mu, scale=sigma)

  ax[0].scatter(x0, pdf_x0, color='darkred', s=100)
  ax[0].axvline(x0, color='darkred', linestyle='-.', lw=0.5)
  ax[0].axhline(pdf_x0, linestyle='-.', lw=0.5, c='darkred')
  ax[0].fill_between(x, pdf_values, where=(x < x0), color='darkred', alpha=0.3)

  ax[1].scatter(x0, cdf_x0, color='darkred', s=100)
  ax[1].axvline(x0, color='darkred', linestyle='-.', lw=0.5)
  ax[1].axhline(cdf_x0, linestyle='-.', lw=0.5, c='darkred')
  # ax[1].fill_between(x, cdf_values, where=(x < x0), color='darkred', alpha=0.3) # não tem significado a área

  ax[2].scatter(prob_x0, ppf_x0, color='darkred', s=100)
  ax[2].axvline(prob_x0, color='darkred', linestyle='-.', lw=0.5)
  ax[2].axhline(ppf_x0, linestyle='-.', lw=0.5, c='darkred')
  # ax[2].fill_between(probabilities, ppf_values, (ax[2].get_ylim()[1] - ax[2].get_ylim()[0])*2, where=(probabilities < prob_x0),  color='darkred', alpha=0.3) # não tem significado a área

  plt.tight_layout()
  plt.show()
```

4. Um exame vestibular tem média de nota dos alunos de 70 pontos, com desvio padrão de 10 pontos. Qual o percentual de alunos com nota entre 50 e 80 pontos? Mostre graficamente como abaixo.

<br>

*Extras... você saberia fazer?*

- Qual a probabilidade de alunos com notas acima de 95 e abaixo de 5 pontos?
- Qual nota de corte deve adotada para selecionar os 75% melhores alunos?

```python
from scipy.stats import norm, t
import numpy as np

fig, ax = plt.subplots(1,2,figsize=(10, 4))

x = np.linspace(40,100,1000)
pdf_values = norm.pdf(x, loc=70, scale=10)
cdfs_values = norm.cdf(x, loc=70, scale=10)

sns.lineplot(x=x, y=pdf_values, label='norm(70,10)',ax=ax[0])
ax[0].vlines(50,0,norm.pdf(50,loc=70,scale=10),linestyle='--',color='k',lw=1)
ax[0].vlines(80,0,norm.pdf(80,loc=70,scale=10),linestyle='--',color='k',lw=1)
ax[0].fill_between(x, pdf_values, where=((x > 50) & (x < 80)), color='red', alpha=0.3)

p = norm.cdf(80,loc=70,scale=10) - norm.cdf(50,loc=70,scale=10)
ax[0].text(70,0.02,f'P = {p:.2f}',ha='center',va='center',fontsize=14)

ax[0].set_title('PDF Plot')
ax[0].legend()

sns.lineplot(x=x, y=cdfs_values, label='norm(70,10)',ax=ax[1])
ax[1].vlines(50,0,norm.cdf(50,loc=70,scale=10),linestyle='--',color='k',lw=1)
ax[1].vlines(80,0,norm.cdf(80,loc=70,scale=10),linestyle='--',color='k',lw=1)
ax[1].hlines(norm.cdf(50,loc=70,scale=10),40,50,linestyle='--',color='k',lw=1)
ax[1].hlines(norm.cdf(80,loc=70,scale=10),40,80,linestyle='--',color='k',lw=1)
ax[1].scatter(50, norm.cdf(50,loc=70,scale=10), color='darkred', s=100)
ax[1].scatter(80, norm.cdf(80,loc=70,scale=10), color='darkred', s=100)
ax[1].set_title('CDF Plot')
ax[1].legend()

plt.tight_layout()
plt.show()
```
